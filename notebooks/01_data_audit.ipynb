## Phase 1 – Data Audit & Structural Engineering

This notebook performs structural validation of all raw datasets.

The purpose of this structural audit is to validate the integrity, granularity, and analytical readiness of all Port Authority datasets prior to any descriptive or exploratory analysis. 

This phase ensures:
- Consistent data types
- Structural completeness
- Reliable temporal alignment
- Dataset-level grain clarity
- Preservation of original analytical characteristics

No analytical transformations were performed during this stage.

```python
# =========================================
# Project Setup
# =========================================

import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)

print("Environment Ready")
```

```python
# =========================================
# Load Raw Datasets
# =========================================

traffic_df = pd.read_csv("../data/raw/All Recorded Traffic.txt", sep="\t")
bus_df = pd.read_csv("../data/raw/All Recorded PABT Bus.txt", sep="\t")
passenger_df = pd.read_csv("../data/raw/All Recorded PABT Passenger.txt", sep="\t")
speeds_df = pd.read_csv("../data/raw/Facility Mobility Speeds.txt", sep="\t")

print("Datasets Loaded Successfully")
```

# 🚦 Traffic Dataset – Structural Audit

```python
# Shape
print("Traffic Shape:", traffic_df.shape)

# Data types
print("\nTraffic Data Types:")
print(traffic_df.dtypes)

# Duplicate rows
print("Traffic Duplicate Rows:", traffic_df.duplicated().sum())

# Missing values
traffic_missing = traffic_df.isna().sum()
print("\nTraffic Missing Values:")
print(traffic_missing[traffic_missing > 0])

# DATE preview
print("\nTraffic DATE Sample:")
print(traffic_df["DATE"].head())
```

### Traffic Dataset – Structural Findings

The Traffic dataset contains 5,383,378 rows and 30 columns with no duplicate records.

Each row represents traffic counts for a specific facility (FAC), lane, date, and time interval. This indicates a highly granular operational dataset at the lane level.

During inspection, the following observations were made:

- Data types were consistent with expected structure.
- All operational traffic flow variables (TOTAL, CASH, EZPASS, Autos, etc.) are structurally complete with zero missing values.
- Date format required conversion to proper datetime format for temporal alignment.
- Missing values (NULL) are concentrated exclusively in the detailed vehicle classification columns (CLASS 1–CLASS 11), with the highest missingness observed in CLASS 6 (1,354,198 records). This indicates inconsistent capture of granular vehicle segmentation while overall traffic totals remain reliable.
- The NULL values were preserved intentionally, as they represent operational gaps rather than structural corruption.
- Column naming conventions include abbreviated identifiers (e.g., FAC, FAC_G2). These were preserved to maintain raw dataset integrity and traceability.

From a structural standpoint, this dataset operates at a highly granular lane-level resolution. 
Any future integration or aggregation will require careful consideration of its temporal and facility-level structure.

```python
# Shape
print("Speeds Shape:", speeds_df.shape)

# Data types
print("\nSpeeds Data Types:")
print(speeds_df.dtypes)

# Duplicate rows
print("\nSpeeds Duplicate Rows:", speeds_df.duplicated().sum())

# Missing values
print("\nSpeeds Missing Values:")
print(speeds_df.isna().sum())

# Speeds Month_Year preview
print("\nSpeeds Month_Year Sample:")
print(speeds_df["Month_Year"].head())
```

### Bus, Passenger & Speeds – Structural Findings

The Bus dataset contains no duplicate records and no missing values. However, the date columns (Start_Date and End_Date) were stored as object type at the time of inspection and therefore required conversion to proper datetime format before analytical processing. Each row represents aggregated bus volume by carrier within a defined date range, indicating a carrier-level aggregation.

The Passenger dataset contains one duplicate record and two structurally incomplete rows identified during inspection. These were flagged for removal prior to descriptive analysis to prevent distortion of summary statistics. Similar to the Bus dataset, its date columns were stored as object type and required conversion to datetime format. The dataset operates at a carrier-level aggregation within defined date ranges.

The Speeds dataset contains no duplicate records and no missing values. The Month_Year column was stored as object type and required conversion to datetime format to ensure proper temporal alignment. Each row represents facility-level mobility metrics at a monthly aggregation.

From a structural perspective, the Bus and Passenger datasets share similar carrier-level aggregation within defined date ranges, making them structurally compatible for future integration after temporal standardization. The Speeds dataset operates at a facility-level monthly aggregation and can be aligned with aggregated Traffic data through facility and time harmonization.
    
No additional structural inconsistencies were identified at this stage. Necessary structural corrections are applied in the following section prior to descriptive and exploratory analysis.

## 🔧 STRUCTURAL CORRECTIONS

```python
# --- Traffic ---
traffic_df["DATE"] = pd.to_datetime(traffic_df["DATE"])

# --- Bus ---
bus_df["Start_Date"] = pd.to_datetime(bus_df["Start_Date"])
bus_df["End_Date"] = pd.to_datetime(bus_df["End_Date"])

# --- Passenger ---
passenger_df["Start_Date"] = pd.to_datetime(passenger_df["Start_Date"])
passenger_df["End_Date"] = pd.to_datetime(passenger_df["End_Date"])

# Remove Passenger duplicate
passenger_df = passenger_df.drop_duplicates()

# Remove Passenger incomplete rows
passenger_df = passenger_df.dropna()

# --- Speeds ---
speeds_df["Month_Year"] = pd.to_datetime(speeds_df["Month_Year"])

print("All structural corrections applied successfully.")
```

```python
# ===============================
# Structural Verification Check
# ===============================

print("Traffic DATE dtype:", traffic_df["DATE"].dtype)
print("Bus Start_Date dtype:", bus_df["Start_Date"].dtype)
print("Bus End_Date dtype:", bus_df["End_Date"].dtype)
print("Passenger Start_Date dtype:", passenger_df["Start_Date"].dtype)
print("Passenger End_Date dtype:", passenger_df["End_Date"].dtype)
print("Speeds Month_Year dtype:", speeds_df["Month_Year"].dtype)

print("\nPassenger Duplicate Rows After Cleaning:", passenger_df.duplicated().sum())
print("Passenger Missing Values After Cleaning:")
print(passenger_df.isna().sum())

# =========================================
# Save Structurally Cleaned Datasets
# =========================================

traffic_df.to_csv("../data/processed/traffic_cleaned.csv", index=False)
bus_df.to_csv("../data/processed/bus_cleaned.csv", index=False)
passenger_df.to_csv("../data/processed/passenger_cleaned.csv", index=False)
speeds_df.to_csv("../data/processed/speeds_cleaned.csv", index=False)

print("Cleaned datasets saved successfully to processed folder.")
```

## Structural Corrections Completed

The following structural changes were applied before starting descriptive analysis and EDA:

- Converted all date-related columns to proper datetime format:
  - Traffic: DATE
  - Bus: Start_Date, End_Date
  - Passenger: Start_Date, End_Date
  - Speeds: Month_Year

This ensures accurate time-based analysis, filtering, and aggregation.

- Removed one duplicate record from the Passenger dataset.
- Removed two incomplete rows from the Passenger dataset.

These adjustments were made because duplicate and incomplete records can distort summary statistics and analytical results.

The missing values in the Traffic dataset (vehicle CLASS columns) were intentionally not treated. These represent analytical missingness in detailed vehicle classification and are part of the original data structure. Modifying them at this stage could alter the analytical characteristics of the dataset.

    At this point, only structural corrections have been performed. No analytical patterns have been modified.The structurally corrected datasets have been saved in the processed directory and will be used for all subsequent descriptive and exploratory analysis. The raw datasets remain unchanged.
