# Phase 4: Sampling

## Description (Progress Recap and Why We Are Sampling)
Across the first three phases of this project, we established a clean and analytically consistent dataset:

- **Phase 1: Data Audit**  
  We verified structural integrity, inspected missingness patterns, and confirmed the data could support downstream analysis.

- **Phase 2: EDA**  
  We explored distributions, trends, and key relationships to understand how traffic and external drivers behave over time and across facilities.

- **Phase 3: Cleaning and Integration**  
  We standardized all datasets to a single analytical grain (**Facility × Date**) and merged the available sources into one daily master table. This created a unified dataset that is structurally stable and ready for modeling.

At this stage, the dataset is large enough that running repeated modeling experiments and diagnostics can become slow and inefficient. Sampling allows us to work faster and iterate safely without changing the “story” the full dataset tells.

The purpose of Phase 4 is therefore to:
1. Select a **representative sample** from the full dataset.
2. Ensure the sample preserves the **same distributional structure** as the original data.
3. Validate that the sample is representative using side‑by‑side statistical comparisons.

> Important: Sampling is used here to accelerate experimentation and analysis. We will verify representativeness so that conclusions drawn from the sample remain consistent with conclusions from the full dataset.


## Process (Sampling Methodology)

### Objective
Create a smaller dataset that remains representative of the full dataset across:
- Facility coverage
- Time coverage (seasonality and long‑term trends)
- Key numeric distributions (e.g., TOTAL traffic, payment channels, vehicle categories, weather variables, passenger and bus volumes)

### Sampling Strategy
We will use **stratified sampling** at the level of:
- **Facility (`FAC`)**
- **Month (`Month_Year` derived from `DATE`)**

This approach is preferred over simple random sampling because it reduces the risk that the sample unintentionally over‑represents:
- specific facilities with higher traffic volumes, or
- specific time periods (e.g., one season or one year)

### Representativeness Validation
After sampling, we will compare the sample to the full dataset using:
1. Side‑by‑side numeric summary statistics (mean, median, standard deviation, range)
2. Distribution comparisons across key grouping dimensions (FAC and Month_Year)
3. Optional statistical tests (KS test for numeric variables)

If the sample distributions closely match the full dataset, the sample will be considered representative and suitable for downstream modeling experiments.


```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv(
    "../data/integrated/final_master_dataset_cleaned.csv",
    parse_dates=["DATE"]
)

print("Full dataset shape:", df.shape)
df.head()
```

## Implementation

In the next steps we will:
1. Create a `Month_Year` column from `DATE` to support time‑stratified sampling.
2. Select a sample fraction (e.g., 10%).
3. Draw a stratified sample within each Facility‑Month group.
4. Validate that the sample matches the full dataset’s key statistics and distribution patterns.


```python
# Month_Year key for stratified sampling (monthly bucket)
df["Month_Year"] = df["DATE"].values.astype("datetime64[M]")

# Helpful for validation (weekday/weekend balance)
df["DayOfWeek"] = df["DATE"].dt.day_name()

print(df[["FAC", "DATE", "Month_Year", "DayOfWeek"]].head())
```

```python
sample_frac = 0.10   # 10% sample (adjust if needed)
seed = 42            # reproducibility

sample = (
    df.groupby(["FAC", "Month_Year"], group_keys=False)
      .apply(lambda g: g.sample(frac=sample_frac, random_state=seed) if len(g) > 1 else g)
)

print("Sample shape:", sample.shape)
sample.head()
```

## Output Interpretation (Sampling Step)

The sample shape confirms the sample was successfully drawn.

Because we sampled proportionally within each Facility‑Month group, the sample should retain:
- coverage across all facilities,
- coverage across the full time range,
- the same seasonal and operational patterns present in the full dataset.

Next, we validate representativeness by comparing key statistics and distributions between the full dataset and the sample.


```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

full_stats = df[numeric_cols].describe().T[["mean", "std", "min", "50%", "max"]]
samp_stats = sample[numeric_cols].describe().T[["mean", "std", "min", "50%", "max"]]

compare = full_stats.join(samp_stats, lsuffix="_full", rsuffix="_sample")

# Percent mean difference (helps quickly spot drift)
compare["mean_diff_%"] = (
    (compare["mean_sample"] - compare["mean_full"]) / compare["mean_full"] * 100
).replace([np.inf, -np.inf], np.nan)

# Show largest absolute differences first
compare_sorted = compare.sort_values("mean_diff_%", key=lambda s: s.abs(), ascending=False)

print(compare_sorted.head(15))
```

```python
def dist_compare(data_full, data_sample, col):
    full = data_full[col].value_counts(normalize=True).rename("full_pct")
    samp = data_sample[col].value_counts(normalize=True).rename("sample_pct")
    out = pd.concat([full, samp], axis=1).fillna(0)
    out["diff_pct_points"] = (out["sample_pct"] - out["full_pct"]) * 100
    return out

print("FAC distribution comparison:")
print(dist_compare(df, sample, "FAC").sort_index())

print("\nDayOfWeek distribution comparison:")
print(dist_compare(df, sample, "DayOfWeek").sort_index())

print("\nMonth_Year distribution comparison (first 12 rows):")
print(dist_compare(df, sample, "Month_Year").sort_index().head(12))
```

```python
from scipy.stats import ks_2samp

ks_results = []
for c in numeric_cols:
    a = df[c].dropna()
    b = sample[c].dropna()
    if a.nunique() > 1 and b.nunique() > 1:
        stat, p = ks_2samp(a, b)
        ks_results.append((c, stat, p))

ks_df = pd.DataFrame(ks_results, columns=["column", "ks_stat", "p_value"]).sort_values("p_value")
print(ks_df.head(20))
```


## What the sample actually showed (Validation Results)

### 1) Facility Representation (FAC) — Preserved Extremely Well
The facility distribution in the sample matches the full dataset almost perfectly. The maximum deviation observed across facilities is approximately **0.09 percentage points**, and most facilities differ by about **0.01 percentage points**.

**Interpretation:** The stratification by `FAC` successfully preserved facility-level representation. This is exactly what we need to avoid facility bias in downstream modeling.

---

### 2) Time / Seasonal Representation (Month_Year) — Preserved Extremely Well
The monthly distribution is also extremely stable. In the Month_Year comparison (example shown for the first 12 months), deviations are typically around **0.01 percentage points**, with the largest shown deviation approximately **0.05 percentage points**.

**Interpretation:** The sampling preserved month-by-month coverage and therefore preserved seasonality and long-run temporal structure at a monthly level.

---

### 3) Day-of-Week Balance — Very Good (Minor Drift)
Day-of-week proportions remain close overall, with a small but noticeable shift:

- **Thursday:** +0.58 percentage points
- **Friday:** −0.51 percentage points  
Other days show smaller deviations.

**Interpretation:** This drift is minor and does not indicate a structural sampling failure. It occurs because the sampling strata were defined as **FAC × Month_Year**, not DayOfWeek. The sample remains credible and usable for modeling, but weekday-sensitive analyses (commuting effects, weekday vs weekend patterns) may show slightly more noise than the full dataset.

---

## Numeric Distribution Similarity (KS Test)
To formally validate distribution similarity for numeric fields, I ran KS tests between the full dataset and the sample.

### Core modeling variables (traffic and volumes) — Strong match
The most important traffic variables show high p-values and low KS statistics, indicating strong distributional alignment:

- **TOTAL:** KS ≈ 0.0102, p ≈ 0.8969  
- Autos, Small_T, Large_T, Buses, CASH, EZPASS, VIOLATION also show very high p-values (generally > 0.80)

**Interpretation:** This is the most important result for predictive modeling: the sample retains the traffic-volume distributions that will drive the predictive signal.

### Weather variables — small but detectable drift
Three weather variables show statistically significant differences:

- **AWND:** KS ≈ 0.0293, p ≈ 0.0088  
- **TAVG:** KS ≈ 0.0254, p ≈ 0.0332  
- **PRCP:** KS ≈ 0.0246, p ≈ 0.0425  

**Interpretation (practical):** The KS statistics are still small (~0.02–0.03). With large datasets, the KS test can flag “significance” even when the real-world magnitude of the difference is small. In practical modeling terms, this indicates **minor drift** in a few weather distributions, not a failure of representativeness.

---

## Overall Examiner Decision: Is the sample representative?
**Yes — the sample is representative and suitable for modeling work.**

- It preserves facility coverage essentially perfectly.
- It preserves monthly time coverage essentially perfectly.
- It preserves core traffic and volume distributions very strongly.
- It shows minor drift in day-of-week proportions and a small detectable drift in a few weather variables, but the effect sizes are small and do not undermine overall representativeness for most modeling tasks.

---

## Is this “perfect”?
It is **not mathematically perfect**, but it is **professionally acceptable and statistically defensible** as a representative sample for:
- rapid modeling iteration,
- AutoML experiments,
- feature engineering,
- pipeline debugging,
- and early model selection.

This sample is strong enough to move forward.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# 1) Choose key numeric columns to compare
# ----------------------------
candidate_numeric = [
    "TOTAL", "CASH", "EZPASS", "VIOLATION",
    "Autos", "Small_T", "Large_T", "Buses",
    "Passenger_Volume", "Bus_Volume",
    "AWND", "PRCP", "SNOW", "TMAX", "TMIN", "TAVG"
]

numeric_cols = [c for c in candidate_numeric if c in df.columns and c in sample.columns]

# ----------------------------
# 2) Helper: overlay density histograms
# ----------------------------
def plot_density_overlay(full_df, samp_df, col, bins=50):
    x_full = full_df[col].dropna()
    x_samp = samp_df[col].dropna()

    if x_full.nunique() < 2 or x_samp.nunique() < 2:
        return  # skip near-constant columns

    # Use common bin edges so shapes are comparable
    lo = min(x_full.min(), x_samp.min())
    hi = max(x_full.max(), x_samp.max())
    edges = np.linspace(lo, hi, bins + 1)

    plt.hist(x_full, bins=edges, density=True, alpha=0.45, label="Full", color="#1f77b4")
    plt.hist(x_samp, bins=edges, density=True, alpha=0.45, label="Sample", color="#ff7f0e")
    plt.title(f"Distribution Comparison: {col}")
    plt.xlabel(col)
    plt.ylabel("Density")
    plt.legend()

# ----------------------------
# 3) Plot numeric distributions (grid)
# ----------------------------
n = len(numeric_cols)
if n == 0:
    print("No overlapping numeric columns found between df and sample.")
else:
    cols_per_row = 3
    rows = int(np.ceil(n / cols_per_row))

    plt.figure(figsize=(cols_per_row * 6, rows * 4))
    for i, col in enumerate(numeric_cols, start=1):
        plt.subplot(rows, cols_per_row, i)
        plot_density_overlay(df, sample, col, bins=60)

    plt.tight_layout()
    plt.show()

# ----------------------------
# 4) Categorical distribution comparison (normalized)
# ----------------------------
def plot_category_distribution(full_df, samp_df, col):
    full_dist = full_df[col].value_counts(normalize=True).sort_index()
    samp_dist = samp_df[col].value_counts(normalize=True).sort_index()

    comp = pd.concat([full_dist.rename("Full"), samp_dist.rename("Sample")], axis=1).fillna(0)
    comp.plot(kind="bar", figsize=(10, 4), color=["#1f77b4", "#ff7f0e"])
    plt.title(f"Categorical Distribution Comparison: {col}")
    plt.ylabel("Proportion")
    plt.tight_layout()
    plt.show()

for cat_col in ["FAC", "DayOfWeek"]:
    if cat_col in df.columns and cat_col in sample.columns:
        plot_category_distribution(df, sample, cat_col)
```

```python
# Monthly mean TOTAL trend: Full vs Sample
if "TOTAL" in df.columns and "Month_Year" in df.columns:
    full_monthly = df.groupby("Month_Year")["TOTAL"].mean()
    samp_monthly = sample.groupby("Month_Year")["TOTAL"].mean()

    plt.figure(figsize=(12, 4))
    plt.plot(full_monthly.index, full_monthly.values, label="Full", color="#1f77b4", linewidth=2)
    plt.plot(samp_monthly.index, samp_monthly.values, label="Sample", color="#ff7f0e", linewidth=2, alpha=0.85)
    plt.title("Monthly Mean TOTAL: Full Dataset vs Sample")
    plt.xlabel("Month_Year")
    plt.ylabel("Mean TOTAL")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## Distribution Charts (Full Dataset vs Sample)

To visually confirm that the sample is representative, I compared the **distributions** of key variables in the full cleaned dataset against the sampled dataset.

Because the full dataset contains far more rows than the sample, I plotted **density-based distributions** (not raw counts). If the sampling is representative, the sample’s curves should closely overlap the full dataset’s curves across the same variables.

In addition to numeric distributions, I also compared the categorical distributions of **FAC** and **DayOfWeek** using normalized bar charts. These charts provide a direct visual confirmation that the sample preserves the same structure observed in the full dataset.

## Interpretation (Distribution Charts)

The overlay density charts show that the sampled dataset closely matches the full cleaned dataset across the primary traffic, vehicle category, and external-driver variables. The two distributions overlap strongly, indicating that sampling preserved the same overall patterns rather than distorting the data.

The normalized bar charts for FAC and DayOfWeek further confirm that categorical structure was preserved. Taken together with the numerical comparisons and statistical validation performed earlier, these visual checks support the conclusion that the sample is representative and appropriate for downstream modeling work.
