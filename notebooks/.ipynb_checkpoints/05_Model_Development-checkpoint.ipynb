### Step 2 — Prepare the dataset for H2O AutoML

After loading and inspecting the sample dataset, the next step is to prepare it for modeling.

At this stage, the goal is to answer Project Question 1 by identifying the factors most strongly associated with total traffic volume. Since `TOTAL` is a continuous numeric variable, this is treated as a regression problem.

Before running H2O AutoML, the dataset must be prepared in a format suitable for modeling:

- `DATE` and `Month_Year` should be converted into date format
- `FAC` and `DayOfWeek` should be treated as categorical variables
- `STATION` will be excluded because it does not add meaningful explanatory value for this question
- the data must be converted into an H2O frame so that AutoML can test multiple regression models

The purpose of this first AutoML run is to allow H2O to compare different regression algorithms and identify the best-performing model for predicting traffic volume.

```python
h2o.init()
```

```python
h2o.init(nthreads=-1, max_mem_size="4G")
```

### Step 6 — Refined predictor selection for meaningful interpretation

At this stage, it is important to ensure that the model provides meaningful business insights rather than mechanically reconstructing the target variable.

Variables such as CASH, EZPASS, VIOLATION, and vehicle category counts (Autos, Small_T, Large_T, Buses) are direct components of TOTAL traffic. Including them would lead to artificially strong model performance without providing actionable insight.

Therefore, for this analysis, these variables are excluded. The model is instead built using:

- facility information  
- day-of-week patterns  
- weather variables  
- passenger and bus-related volumes  

This ensures that the resulting model identifies real external drivers of traffic rather than simply decomposing the total.

```python
question1_data = traffic_data.copy()

# Drop STATION because it is not useful for this modeling question
question1_data = question1_data.drop(columns=["STATION"])

# Convert date columns
question1_data["DATE"] = pd.to_datetime(question1_data["DATE"])
question1_data["Month_Year"] = pd.to_datetime(question1_data["Month_Year"])

print("Shape after preparation:", question1_data.shape)
question1_data.head()
```

```python
question1_h2o = h2o.H2OFrame(question1_data)

question1_h2o["FAC"] = question1_h2o["FAC"].asfactor()
question1_h2o["DayOfWeek"] = question1_h2o["DayOfWeek"].asfactor()
```

```python
target_column = "TOTAL"

predictor_columns = [
    "FAC",
    "AWND",
    "PRCP",
    "SNOW",
    "TMAX",
    "TMIN",
    "TAVG",
    "Passenger_Volume",
    "Bus_Volume",
    "DayOfWeek"
]

print("Target variable:", target_column)
print("Predictors:", predictor_columns)
```

```python
train_q1, test_q1 = question1_h2o.split_frame(ratios=[0.8], seed=42)

print("Training rows:", train_q1.nrows)
print("Testing rows:", test_q1.nrows)
```

```python
automl_q1 = H2OAutoML(
    max_models=20,
    seed=42,
    sort_metric="RMSE",
    nfolds=5
)

automl_q1.train(
    x=predictor_columns,
    y=target_column,
    training_frame=train_q1,
    leaderboard_frame=test_q1
)
```

```python
leaderboard_q1 = automl_q1.leaderboard
leaderboard_q1.head(rows=20)
best_model_q1 = automl_q1.leader
best_model_q1
```

```python
best_model_q1.varimp(use_pandas=True)
```

```python
best_model_q1.explain(test_q1)
```

```python
leaderboard_q1_df = automl_q1.leaderboard.as_data_frame()
leaderboard_q1_df
```

```python
non_ensemble_models = leaderboard_q1_df[
    ~leaderboard_q1_df["model_id"].str.contains("StackedEnsemble", case=False, na=False)
]

non_ensemble_models.head(10)
```

```python
best_interpretable_model_id = non_ensemble_models.iloc[0]["model_id"]
best_interpretable_model = h2o.get_model(best_interpretable_model_id)

print("Best interpretable model ID:", best_interpretable_model_id)
best_interpretable_model
```

```python
interpretable_performance_q1 = best_interpretable_model.model_performance(test_q1)
interpretable_performance_q1
```

```python
variable_importance_q1 = best_interpretable_model.varimp(use_pandas=True)
variable_importance_q1.head(10)
```

### Step 11 — Identifying the key drivers of traffic (Final Answer)

To answer the question “What are the top factors affecting traffic?”, we analyzed the variable importance from the best-performing interpretable model (GBM) identified by H2O AutoML.

While the stacked ensemble model provided the highest predictive accuracy, it is less suitable for direct interpretation. Therefore, the highest-ranked non-ensemble model was used to extract variable importance and identify key drivers.

The analysis revealed the following top five factors influencing traffic volume:

1. Facility (FAC)  
2. Day of Week  
3. Maximum Temperature (TMAX)  
4. Average Temperature (TAVG)  
5. Wind Speed (AWND)  

The results indicate that traffic patterns are primarily driven by structural and temporal factors rather than purely demand-based variables.

Facility (FAC) emerged as the most important factor, suggesting that traffic volume varies significantly across different bridges and tunnels. This reflects differences in geographic location, connectivity, and usage patterns across Port Authority infrastructure.

Day-of-week effects further highlight predictable commuting behavior, with clear distinctions between weekday and weekend traffic patterns.

Weather variables, particularly temperature and wind, also play a meaningful role. Higher or lower temperatures influence travel demand, while wind conditions can affect driving comfort and traffic flow, especially on exposed infrastructure such as bridges.

Interestingly, passenger and bus-related variables were less influential than expected, suggesting that overall traffic volume is more strongly determined by infrastructure and environmental conditions than by passenger movement alone.

These findings provide a strong foundation for operational planning, as they identify the key external factors that Port Authority should monitor and incorporate into traffic management strategies.

The objective is not to predict violations, but to quantify them and identify patterns across time and location. By doing so, the Port Authority can determine high-risk facilities and time periods where enforcement and operational interventions are most needed.

### Models and Tools Used

Unlike Question 1, this is not a predictive modeling problem. The business requirement is to compute and analyze counts rather than forecast or classify outcomes. Therefore, regression, classification, or clustering models are not appropriate for this task.

Instead, we use descriptive analytics techniques implemented in Python, including:
- Data aggregation using groupby operations
- Time-based feature engineering (year, month, week)
- Facility-level breakdown analysis
- Visualization for pattern identification

This approach ensures that the results are transparent, directly aligned with the business question, and easy to interpret for decision-making.

### Why a model is NOT used here

Using a model in this case would not add value, because the company is not asking for predictions or classifications. Instead, the goal is to measure actual violation counts and understand their distribution. Applying AutoML or predictive models would introduce unnecessary complexity without improving the accuracy or usefulness of the answer.

### What we are doing

To answer this question, we aggregate violation counts across:
- Time intervals (year, month, week)
- Facilities (bridges and tunnels)

This allows us to identify:
- overall volume of toll violations
- temporal trends (growth, seasonality)
- facility-level concentration of violations
- operational patterns relevant for enforcement

### Why we are doing this

Understanding when and where violations occur helps the Port Authority:
- reduce revenue leakage
- prioritize enforcement resources
- identify high-risk facilities
- improve operational and policy decisions

```python
# Convert DATE to datetime
traffic_data["DATE"] = pd.to_datetime(traffic_data["DATE"])

# Create additional time features
traffic_data["Year"] = traffic_data["DATE"].dt.year
traffic_data["Month"] = traffic_data["DATE"].dt.month
traffic_data["Week"] = traffic_data["DATE"].dt.isocalendar().week
```

```python
total_violations = traffic_data["VIOLATION"].sum()
print("Total Violations:", total_violations)
```

#### Total Violations

The analysis shows that:

👉 **Total Toll Violations = 8,023,030**

This indicates that toll violations are a **large-scale operational issue**, not isolated incidents. The magnitude suggests that even small improvements in enforcement could result in significant revenue recovery.

---

```python
violations_by_year = traffic_data.groupby("Year")["VIOLATION"].sum().reset_index()

import matplotlib.pyplot as plt

# Create yearly summary
violations_by_year = (
    traffic_data.groupby("Year")["VIOLATION"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 5))
plt.plot(
    violations_by_year["Year"],
    violations_by_year["VIOLATION"],
    marker="o",
    linewidth=2
)

# Add value labels
for x, y in zip(violations_by_year["Year"], violations_by_year["VIOLATION"]):
    plt.text(
        x,
        y + 10000,   # 🔥 add a vertical buffer (adjust based on scale)
        f"{y:,}",
        ha="center",
        fontsize=9
    )

plt.title("Toll Violations by Year", fontsize=14, pad=12)
plt.xlabel("Year", fontsize=11)
plt.ylabel("Number of Violations", fontsize=11)
plt.xticks(violations_by_year["Year"], rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

From the yearly trend:

- Violations increased steadily from **2013 (~290K)** to **2020 (~1.25M)**
- Peak occurred around **2020**
- A decline is observed after 2021

**Interpretation:**

- The steady increase suggests **growing system exploitation over time**
- The spike around 2020 may reflect:
  - changes in travel behavior
  - enforcement gaps
  - increased dependence on toll infrastructure
- The decline after 2022 may be due to:
  - partial data
  - policy or enforcement improvements

👉 Key Insight:  
**Violations follow long-term trends and are not random events**


```python
violations_by_month = traffic_data.groupby("Month")["VIOLATION"].sum().reset_index()
# Create monthly summary
violations_by_month = (
    traffic_data.groupby("Month")["VIOLATION"]
    .sum()
    .reset_index()
)

# Month labels
month_labels = {
    1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr",
    5: "May", 6: "Jun", 7: "Jul", 8: "Aug",
    9: "Sep", 10: "Oct", 11: "Nov", 12: "Dec"
}

violations_by_month["Month_Name"] = violations_by_month["Month"].map(month_labels)

plt.figure(figsize=(11, 5))
bars = plt.bar(
    violations_by_month["Month_Name"],
    violations_by_month["VIOLATION"]
)

# Add labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.annotate(
        f"{int(height):,}",
        (bar.get_x() + bar.get_width() / 2, height),
        textcoords="offset points",
        xytext=(0, 6),
        ha="center",
        fontsize=8
    )

plt.title("Toll Violations by Month", fontsize=14, pad=12)
plt.xlabel("Month", fontsize=11)
plt.ylabel("Number of Violations", fontsize=11)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

#### Violations by Month (Seasonality)

Monthly analysis shows:

- Higher violations during **May–August**
- Lower violations during **January–February**

**Interpretation:**

- Warmer months → increased travel → more violations  
- Winter months → reduced travel → fewer violations  

👉 Key Insight:  
**Violations are strongly correlated with traffic volume and seasonality**


```python
violations_by_week = traffic_data.groupby("Week")["VIOLATION"].sum().reset_index()
violations_by_week.head()
```

```python
violations_by_fac = traffic_data.groupby("FAC")["VIOLATION"].sum().reset_index()

# Create facility summary and sort descending
violations_by_fac = (
    traffic_data.groupby("FAC")["VIOLATION"]
    .sum()
    .reset_index()
    .sort_values(by="VIOLATION", ascending=False)
)

plt.figure(figsize=(10, 5))
bars = plt.bar(
    violations_by_fac["FAC"].astype(str),
    violations_by_fac["VIOLATION"]
)

# Add labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.annotate(
        f"{int(height):,}",
        (bar.get_x() + bar.get_width() / 2, height),
        textcoords="offset points",
        xytext=(0, 6),
        ha="center",
        fontsize=9
    )

plt.title("Toll Violations by Facility", fontsize=14, pad=12)
plt.xlabel("Facility", fontsize=11)
plt.ylabel("Number of Violations", fontsize=11)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Ensure DATE is datetime
traffic_data["DATE"] = pd.to_datetime(traffic_data["DATE"])

# Create time-based features
traffic_data["Year"] = traffic_data["DATE"].dt.year
traffic_data["Month"] = traffic_data["DATE"].dt.month
traffic_data["Week"] = traffic_data["DATE"].dt.isocalendar().week
traffic_data["Day"] = traffic_data["DATE"].dt.day
```

```python
# Traffic by Year
traffic_by_year = traffic_data.groupby("Year")["TOTAL"].sum()

print(traffic_by_year)

traffic_by_year.plot(title="Total Traffic by Year", marker='o')
```


The yearly trend shows that traffic increased steadily from 2013 through 2019, rising from approximately 11.2 million to over 12.2 million vehicles. This indicates a **consistent growth in demand** for bridge and tunnel usage over time.

A sharp decline is observed in 2020, where traffic drops significantly to around 9.6 million. This represents a **major external disruption**, rather than a normal fluctuation. Traffic then recovers in the following years, reaching over 12.1 million again by 2023. The lower value in 2025 reflects partial-year data rather than a true decline.

Overall, this shows that traffic demand is **structurally increasing over time**, while remaining sensitive to external shocks.

```python
# Traffic by Month
traffic_by_month = traffic_data.groupby("Month")["TOTAL"].mean()

print(traffic_by_month)

traffic_by_month.plot(title="Average Traffic by Month", marker='o')
```


The monthly analysis reveals a clear **seasonal pattern**. Traffic is lowest in the early months of the year, particularly January and February, and gradually increases toward the summer months.

Peak traffic occurs between **June and August**, where average daily traffic exceeds 44,000 vehicles. After August, traffic begins to decline toward the end of the year.

This pattern indicates that traffic is strongly influenced by **seasonality**, with higher travel demand during warmer months and lower demand during colder periods.

```python
# Traffic by Day of Week
traffic_by_day = traffic_data.groupby("DayOfWeek")["TOTAL"].mean().sort_values(ascending=False)

print(traffic_by_day)

traffic_by_day.plot(kind="bar", title="Traffic by Day of Week")
```


The day-of-week analysis shows that traffic is relatively stable throughout the week but follows a consistent pattern. Traffic is slightly higher toward the end of the week, particularly on **Fridays and Saturdays**, and lowest on Tuesdays.

This reflects a combination of **commuting behavior and leisure travel**, where weekday traffic is driven by work-related movement and weekend traffic includes discretionary travel.

```python
# Weekday vs Weekend
traffic_data["Is_Weekend"] = traffic_data["DayOfWeek"].isin(["Saturday", "Sunday"])

weekday_weekend = traffic_data.groupby("Is_Weekend")["TOTAL"].mean()
weekday_weekend.index = ["Weekday", "Weekend"]

print(weekday_weekend)
```


The comparison between weekday and weekend traffic shows that **weekend traffic (~42,056)** is slightly higher than **weekday traffic (~41,110)**.

This difference is not large, but it is meaningful. It suggests that traffic is not driven exclusively by commuting patterns. Instead, **leisure and discretionary travel during weekends** plays an important role in overall traffic levels.

At the same time, the relatively small gap confirms that **commuting demand remains a strong contributor**, resulting in a balanced traffic structure across the week.

```python
# Busiest Days
busiest_days = traffic_data.sort_values(by="TOTAL", ascending=False).head(10)

busiest_days["DayOfWeek"] = busiest_days["DATE"].dt.day_name()

print(busiest_days[["DATE", "TOTAL", "FAC", "DayOfWeek"]])
```

The analysis of the top 10 busiest days shows that peak traffic events are **highly concentrated in specific facilities**, particularly Facility 7.

The maximum observed traffic exceeds **95,000 vehicles in a single day**, which is significantly higher than average levels. These peak days are not randomly distributed and tend to occur during **high-demand periods**.

This indicates that extreme congestion is **location-specific and event-driven**, rather than uniform across all facilities.

```python
# Correlation
traffic_violation_corr = traffic_data[["TOTAL", "VIOLATION"]].corr()

print(traffic_violation_corr)
```

The correlation between traffic and violations is approximately **0.48**, indicating a **moderate positive relationship**.

This suggests that as traffic increases, violations also tend to increase. Higher traffic volumes create more opportunities for toll evasion, leading to increased enforcement pressure during peak periods.

```python
vehicle_cols = ["Autos", "Small_T", "Large_T", "Buses"]

vehicle_corr = traffic_data[vehicle_cols + ["TOTAL"]].corr()

print(vehicle_corr)
```

Passenger vehicles (**Autos**) show an extremely strong correlation with total traffic (~0.98), indicating that they are the **primary driver of traffic variation**.

Trucks contribute moderately, while buses have a smaller impact. This suggests that congestion is largely driven by **passenger vehicle flow**.

```python
monthly_analysis = traffic_data.groupby("Month")[["TOTAL", "VIOLATION"]].sum()

print(monthly_analysis)

monthly_analysis.plot(title="Traffic vs Violations by Month")
```


Both traffic and violations follow similar patterns across months, with peaks during **late spring and summer**.

This reinforces the conclusion that violations are **partly driven by traffic volume**, rather than occurring independently.

```python
weather_cols = ["AWND", "PRCP", "SNOW", "TMAX", "TMIN", "TAVG"]

weather_corr = traffic_data[weather_cols + ["TOTAL"]].corr()

print(weather_corr["TOTAL"])

monthly_weather = traffic_data.groupby("Month")[["TOTAL", "TAVG"]].mean()

# Normalize
normalized = (monthly_weather - monthly_weather.min()) / (monthly_weather.max() - monthly_weather.min())

normalized.plot(title="Normalized Traffic vs Temperature by Month")
```



The normalized comparison between traffic and average temperature shows that both variables follow a similar seasonal pattern. Traffic and temperature increase together from the beginning of the year, peak during the summer months, and decline toward the end of the year.

This indicates a clear directional relationship, where warmer conditions are associated with higher traffic levels. However, the statistical correlation between temperature and traffic is relatively weak (approximately 0.11–0.12), suggesting that temperature alone does not strongly explain variations in traffic.

This apparent difference between the visual trend and the correlation result highlights that temperature is closely aligned with broader seasonal patterns rather than acting as an independent driver of traffic. In other words, weather reinforces seasonality but is not the primary factor influencing traffic behavior.

Additionally, adverse weather conditions such as snow show a negative relationship with traffic, indicating reduced travel during such periods.

Overall, weather plays a supporting role in traffic patterns, while time-based factors such as seasonality and travel behavior remain the dominant drivers.

## Project Question 4  

### What are we analyzing?

In this section, we evaluate whether congestion pricing has led to a **shift in traffic patterns across facilities in 2025**.  

The goal is to determine whether drivers are **changing their route choices**, potentially avoiding certain facilities and increasing congestion in others.

### Why this matters

Understanding how traffic redistributes across facilities allows us to:

- Detect **behavioral shifts caused by pricing or congestion**
- Identify facilities experiencing **increased or reduced demand**
- Support **data-driven traffic management and policy decisions**

### Analytical approach

This analysis is conducted in two parts:

**1. Descriptive Analysis (Primary Focus)**  
We compare traffic patterns before 2025 and during 2025 using **facility-level traffic share** to identify shifts in driver behavior.

**2. Model-Based Analysis (Supporting)**  
We build a classification model using H2O AutoML to identify the key factors associated with **high traffic (congestion conditions)** and support the observed patterns.

### Key idea

If traffic share increases in some facilities and decreases in others, it indicates that drivers are **rerouting**, which provides direct evidence of congestion or pricing impact.

The model further helps validate and explain the conditions under which congestion occurs.

```python
# -------------------------------
# STEP 1: Ensure datetime + extract year
# -------------------------------
traffic_data["DATE"] = pd.to_datetime(traffic_data["DATE"])
traffic_data["Year"] = traffic_data["DATE"].dt.year

# -------------------------------
# STEP 2: Split data (Before vs 2025)
# -------------------------------
before_2025 = traffic_data[traffic_data["Year"] < 2025]
data_2025 = traffic_data[traffic_data["Year"] == 2025]

# -------------------------------
# STEP 3: Traffic by facility
# -------------------------------
before_fac = (
    before_2025.groupby("FAC")["TOTAL"]
    .sum()
    .reset_index()
)

after_fac = (
    data_2025.groupby("FAC")["TOTAL"]
    .sum()
    .reset_index()
)

# -------------------------------
# STEP 4: Convert to share (%)
# -------------------------------
before_fac["Share_Before"] = before_fac["TOTAL"] / before_fac["TOTAL"].sum() * 100
after_fac["Share_2025"] = after_fac["TOTAL"] / after_fac["TOTAL"].sum() * 100

# Merge
comparison = before_fac.merge(after_fac, on="FAC", suffixes=("_Before", "_2025"))

# Change in share
comparison["Change_%"] = comparison["Share_2025"] - comparison["Share_Before"]

# Sort descending (most increase first)
comparison = comparison.sort_values(by="Change_%", ascending=False)

print("Traffic Share Comparison (Before vs 2025):")
print(comparison)

# -------------------------------
# STEP 5: Visualization
# -------------------------------
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    comparison["FAC"].astype(str),
    comparison["Change_%"]
)

plt.title("Change in Traffic Share by Facility (2025 vs Before)")
plt.xlabel("Facility")
plt.ylabel("Change in Share (%)")

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```


### What we observe

The comparison of traffic share across facilities clearly shows that traffic distribution has shifted in 2025 relative to previous years.

The most significant increase is observed in **Facility 7 (+2.94%)**, followed by **Facility 5 (+1.43%)**, **Facility 9 (+0.87%)**, and **Facility 4 (+0.59%)**.  

On the other hand, facilities such as **Facility 2 (-0.71%)**, **Facility 1 (-0.30%)**, and **Facility 6 (-0.28%)** show a decline in traffic share.

### Key interpretation

These changes indicate that traffic is **not evenly distributed across facilities** and that drivers are adjusting their route choices.

The increase in traffic share in certain facilities suggests that drivers may be:
- **Rerouting toward alternative crossings**
- Avoiding specific facilities that may be more congested or costly  

At the same time, the decline in other facilities indicates potential **avoidance behavior**, where drivers are shifting away from those routes.

### Why this matters

This shift is critical because it shows that congestion pricing or congestion-related factors do not reduce traffic uniformly. Instead, traffic is **redistributed across the network**.

This creates a situation where:
- Some facilities experience **reduced pressure**
- Others experience **increased congestion**

### What this means for congestion

The strong increase in Facility 7 suggests that it is absorbing a larger portion of total traffic and may face **higher congestion risk**.

Similarly, facilities with declining share may be experiencing **reduced demand**, possibly due to pricing sensitivity or congestion avoidance.

This confirms that congestion is:
- **Dynamic**
- **Facility-specific**
- Influenced by driver decision-making  

### Conclusion

The results provide clear evidence that traffic patterns in 2025 differ from historical patterns.

The observed redistribution across facilities indicates that drivers are **adapting their behavior**, likely in response to congestion or pricing conditions.

Rather than eliminating congestion, these changes suggest that congestion is being **shifted across facilities**, which has important implications for traffic management and policy design.

To further support these findings, a classification model is developed in the next step to identify the key factors associated with high traffic (congestion conditions). This helps validate whether the observed shifts are consistent with underlying traffic drivers such as facility, time patterns, and environmental conditions.


### Model-Based Analysis: Identifying Drivers of Congestion

After identifying a shift in traffic patterns across facilities, we now build a model to understand the **underlying drivers of congestion**.

### Why we are building a model

The descriptive analysis shows that traffic is being redistributed across facilities.  

However, to fully understand this behavior, we need to identify:

- What conditions lead to **high traffic (congestion)**
- Whether congestion is influenced by **facility, time, or external factors**
- Whether the observed shift aligns with known traffic drivers  

This allows us to move from **observation → explanation**.


### Modeling approach

We treat congestion as a **classification problem**, where:

- High Traffic (1) → Congestion condition  
- Normal Traffic (0) → Non-congested condition  

### Key decisions

**1. Target Variable**  
We define congestion using the **75th percentile of traffic (TOTAL)**.  
This ensures we focus on **peak traffic conditions**, rather than arbitrary thresholds.

**2. Model Selection (AutoML)**  
We use H2O AutoML because:

- It evaluates multiple algorithms (GBM, Random Forest, Deep Learning, etc.)
- It selects the best-performing model objectively  
- It avoids bias in model selection  

**3. Avoiding Data Leakage**  
We exclude variables such as:

- TOTAL  
- CASH  
- EZPASS  
- VIOLATION  
- Vehicle counts  

These are direct components of traffic and would artificially inflate model performance.

**4. Features Used**  

We include:

- Facility (FAC)  
- Time patterns (DayOfWeek)  
- Weather variables (TAVG, PRCP, SNOW, etc.)  
- Passenger and bus volumes  

These represent **true drivers of traffic behavior**, not derived outcomes.

### Objective

The goal of this model is not just prediction, but to identify:

- Which factors most strongly influence congestion  
- Whether facility-level patterns align with observed traffic shifts  

```python
# -------------------------------
# STEP 1: Create target variable
# -------------------------------
threshold = traffic_data["TOTAL"].quantile(0.75)

traffic_data["High_Traffic"] = (traffic_data["TOTAL"] >= threshold).astype(int)

print("High Traffic Threshold:", threshold)
traffic_data[["TOTAL", "High_Traffic"]].head()


# -------------------------------
# STEP 2: Prepare dataset
# -------------------------------
question4_data = traffic_data.copy()

# Drop unnecessary columns
if "STATION" in question4_data.columns:
    question4_data = question4_data.drop(columns=["STATION"])

# Convert date fields
question4_data["DATE"] = pd.to_datetime(question4_data["DATE"])
question4_data["Month_Year"] = pd.to_datetime(question4_data["Month_Year"])
```

```python
# -------------------------------
# STEP 3: Convert to H2O frame
# -------------------------------
import h2o
from h2o.automl import H2OAutoML

# (Safe — won't restart if already running)
h2o.init()

question4_h2o = h2o.H2OFrame(question4_data)

# Convert categorical variables
question4_h2o["FAC"] = question4_h2o["FAC"].asfactor()
question4_h2o["DayOfWeek"] = question4_h2o["DayOfWeek"].asfactor()
question4_h2o["High_Traffic"] = question4_h2o["High_Traffic"].asfactor()


# -------------------------------
# STEP 4: Define target & predictors
# -------------------------------
target_q4 = "High_Traffic"

predictors_q4 = [
    "FAC",
    "AWND",
    "PRCP",
    "SNOW",
    "TMAX",
    "TMIN",
    "TAVG",
    "Passenger_Volume",
    "Bus_Volume",
    "DayOfWeek"
]


# -------------------------------
# STEP 5: Train-test split
# -------------------------------
train_q4, test_q4 = question4_h2o.split_frame(ratios=[0.8], seed=42)

print("Train rows:", train_q4.nrows)
print("Test rows:", test_q4.nrows)


# -------------------------------
# STEP 6: Run AutoML
# -------------------------------
automl_q4 = H2OAutoML(
    max_models=15,
    seed=42,
    nfolds=5,
    sort_metric="AUC"
)

automl_q4.train(
    x=predictors_q4,
    y=target_q4,
    training_frame=train_q4,
    leaderboard_frame=test_q4
)
```

```python
# -------------------------------
# STEP 7: Leaderboard
# -------------------------------
leaderboard_q4 = automl_q4.leaderboard
leaderboard_q4.head(rows=15)
```

```python
# -------------------------------
# STEP 8: Best model
# -------------------------------
best_model_q4 = automl_q4.leader
best_model_q4
```

```python
# -------------------------------
# STEP 9: Get interpretable model
# -------------------------------
leaderboard_df_q4 = leaderboard_q4.as_data_frame()

non_ensemble_q4 = leaderboard_df_q4[
    ~leaderboard_df_q4["model_id"].str.contains("StackedEnsemble", case=False, na=False)
]


best_model_id_q4 = non_ensemble_q4.iloc[0]["model_id"]

best_interpretable_q4 = h2o.get_model(best_model_id_q4)

print("Best interpretable model:", best_model_id_q4)

```

```python
# -------------------------------
# STEP 10: Variable importance
# -------------------------------
varimp_q4 = best_interpretable_q4.varimp(use_pandas=True)
varimp_q4.head(10)
```

### Model performance

The AutoML results show that the best-performing model is a **Stacked Ensemble**, which combines multiple algorithms to achieve higher predictive accuracy.

The model achieves an **AUC of approximately 0.976–0.984**, indicating excellent ability to distinguish between high traffic (congestion) and normal traffic conditions.  

The overall accuracy is approximately **93%**, with a low error rate (~7%), confirming that congestion patterns can be reliably identified using the selected features.

### Model interpretation

Although the Stacked Ensemble provides the best performance, we focus on the best non-ensemble model (GBM) for interpretability.

The variable importance results show that:

- **Facility (FAC)** is the dominant factor, contributing nearly **88% of the model importance**
- Time-related patterns such as **DayOfWeek** also play a meaningful role  
- Weather variables (TAVG, TMAX, TMIN) have smaller but noticeable contributions  
- Passenger and bus volumes have relatively minor impact  

### Key insight from the model

The dominance of the **facility variable** confirms that congestion is primarily **location-driven**, rather than being evenly distributed across the system.

This aligns directly with the descriptive analysis, where we observed that certain facilities (such as Facility 7) are gaining traffic share, while others are losing it.

### Connection to traffic shift

The model reinforces the earlier findings:

- Facilities gaining traffic share are also associated with **higher likelihood of congestion**
- This indicates that traffic is not disappearing, but being **redistributed across facilities**
- Driver behavior is adapting, likely in response to congestion or pricing conditions  

### Why AutoML was used

We used AutoML to:

- Evaluate multiple models objectively  
- Select the best-performing algorithm without bias  
- Ensure robustness and reliability in the results  

This approach ensures that the conclusions are **data-driven rather than assumption-based**.

### Conclusion from model

The model confirms that congestion is:

- **Highly predictable**
- **Primarily driven by facility-level characteristics**
- Influenced by time patterns and, to a lesser extent, environmental factors  

This provides strong evidence that the observed traffic shifts are **not random**, but are driven by underlying structural and behavioral factors.


The analysis shows that traffic patterns in 2025 have shifted across facilities rather than remaining consistent with historical behavior.

The descriptive results clearly indicate that some facilities are gaining traffic share while others are losing it, confirming that traffic is being redistributed across the system.

The model further reinforces that congestion is primarily driven by facility-level characteristics, with time patterns and environmental factors playing supporting roles.

## Project Question 5  

What is the expected future usage of bridges and tunnels beyond 2025?


The fifth question moves from analyzing past traffic patterns to forecasting future demand across facilities. The objective is to estimate how traffic usage will evolve over time and support long-term planning decisions for infrastructure, congestion management, and pricing strategies.

This is fundamentally a time-series forecasting problem. Unlike previous questions, where the goal was to explain relationships or identify patterns, here the focus is on predicting future values based on historical behavior. Traffic data is inherently time-dependent, meaning that past trends, seasonal effects, and structural changes influence future outcomes.


### Modeling Approach

We initially attempted a simple time-series model using aggregated traffic data. However, this resulted in a flat forecast, indicating that the model was not capturing meaningful variation in the data. This occurs when important structural information is lost due to excessive aggregation.

To address this limitation, we refined the modeling approach.

Instead of forecasting total traffic as a single series, we implemented **facility-level time-series forecasting**, where each facility is treated as a separate time series. This approach allows the model to capture:

- Differences in traffic behavior across facilities  
- Seasonal patterns at a finer level  
- Structural variations in demand  

### Why this approach is better

Forecasting at the facility level is more aligned with how the Port Authority operates. Traffic demand is not uniform across the system; it varies significantly by location, infrastructure role, and connectivity. By modeling each facility separately, we preserve this variation and generate more meaningful forecasts.

This also allows the Authority to make **facility-specific decisions**, rather than relying on a single system-wide estimate.


### Data Preparation

Traffic data was converted into a time-series format using monthly aggregation. Monthly data was selected instead of yearly data because it provides sufficient granularity to capture seasonality while maintaining stability in the model.

Each facility is treated as an independent time series, with:

- Time variable: Date (monthly)
- Target variable: TOTAL traffic
- Series identifier: FAC (facility)


### Objective

The goal of this model is to:

- Forecast future traffic usage for each facility  
- Identify long-term trends in demand  
- Support strategic planning and congestion management decisions  

```python
# -------------------------------
# STEP 1: Import library
# -------------------------------

from autots import AutoTS
import pandas as pd
import matplotlib.pyplot as plt

```

```python
# -------------------------------
# STEP 2: Prepare data
# -------------------------------
traffic_data["DATE"] = pd.to_datetime(traffic_data["DATE"])

# Monthly aggregation per facility
monthly_facility = (
    traffic_data.groupby(
        [traffic_data["DATE"].dt.to_period("M"), "FAC"]
    )["TOTAL"]
    .sum()
    .reset_index()
)

# Convert period → timestamp
monthly_facility["DATE"] = monthly_facility["DATE"].dt.to_timestamp()

# Rename columns for AutoTS
monthly_facility = monthly_facility.rename(columns={
    "DATE": "datetime",
    "TOTAL": "value",
    "FAC": "series_id"
})

print(monthly_facility.head())
```

```python
# -------------------------------
# STEP 3: Initialize AutoTS
# -------------------------------
model = AutoTS(
    forecast_length=24,       # 2 years forecast
    frequency='M',
    ensemble='simple',
    max_generations=5,
    num_validations=2,
    model_list="fast",
    transformer_list="fast"
)
```

```python
# -------------------------------
# STEP 4: Fit model
# -------------------------------
model = model.fit(
    monthly_facility,
    date_col='datetime',
    value_col='value',
    id_col='series_id'
)
```

```python
# -------------------------------
# STEP 5: Forecast
# -------------------------------
prediction = model.predict()

forecast = prediction.forecast

print("Forecast:")
print(forecast.head())
```

```python
# STEP 6: Plot (example facility)
# -------------------------------
# Plot for one facility (e.g., FAC = 7)

facility_7 = monthly_facility[monthly_facility["series_id"] == 7]

forecast_7 = forecast[7]

plt.figure(figsize=(10,5))

plt.plot(facility_7["datetime"], facility_7["value"], label="Actual")
plt.plot(forecast_7.index, forecast_7.values, label="Forecast")

plt.title("Facility 7 Traffic Forecast")
plt.xlabel("Date")
plt.ylabel("Traffic")

plt.legend()
plt.grid(True)

plt.show()

```
